#Introduction to Python

```python
a = 16
b = 3

print("Sum (a+b) =", a+b)
print("Difference (a-b) =", a-b)
print("Absolute Difference (abs(a-b)) =", abs(a-b))
print("Product (a*b) =", a*b)
print("Floor Division (a//b) =", a//b)
print("Quotient (round(a/b, 4)) =", round(a/b, 4))
print("Reminder (a%b) =", a%b)
print("Exponent (a**b) =", a**b)

x = input("Enter two numbers: ")
y = input()
print(f'x is {x}, y is {y}')
print("x==y:", x==y)
print("x!=y:", x!=y)
```

#Strings

```python
str = "This Is A String"
s2 = "This is a string \nwritten in two lines"

print(str)
print(s2)

print(str[:4:])
print(str[6:])
print(str[::3])
print(str[::-1])
print(str[13:2:-2])
```

Methods and Properties

```python
print(str.upper())
print(str.lower())

a = "Red Flag"
print("'a' is", a)
a.replace("Red", "Green")
print("'a' changed to", a)

b = "THIS, IS A, STRING !!!"
print("'b' is", b)
print(b.split())
print(b.split(","))
print(b.split("!"))
```

Concatination

```python
print("Concatination of strings 'a' and 'b'")
print(a + "... " + b)
```

String formatting

```python
name = "Keerthana"
greet = f"Hi, my name is {name}"
print(greet)
```

#Lists

```python
list_1 = [2,17,18,19,20]
list_2 = ["Keerthana", 21, "Sistla", 101]
list_3 = ["XYZ", ["ABC", 7], [1, "PQR", 0.5]]
```

List Indexing

```python
print(list_1)
print(list_1[3])
print(list_1[-3])
print(list_1[:3:2])
print(list_1[2:5])
```

Element Type

```python
print(list_2)
print(type(list_2))
print(type(list_2[2]))
list_2[2] = 15
print(list_2)
print(list_2[2])
print(type(list_2[2]))
```

List Mutability

```python
print(list_3)
print(list_3[2][1])
list_3[2][1]=13
print(list_3[2][1])
print(list_3)
```

Methods

```python
list_1.append(list_2)
print(list_1)
list_1.extend(list_2)
print(list_1)
```

```python
list_1.insert(2, 1000)
print(list_1)
```

```python
list_1.remove(20)
print(list_1)
removed_item = list_1.pop()
print(removed_item)
print(type(removed_item))
```

```python
list_1 = [2, 17, 1000, 18, 19, 21, 15]
```

```python
print("Len:", len(list_1))
print("Max:", max(list_1))
print("Min:", min(list_1))
```

#Dictionaries

```python
dict_1 = {'Brand':'Ford', "Cost":25000}
print(dict_1)
```

```python
print(dict_1['Brand'])
```

```python
dict_1['Brand']="Audi"
print(dict_1['Brand'])
print(dict_1)
```

```python
dict_1['Transmission']='Manual'
print(dict_1)
```

```python
dict_1['other_attributes'] = {'color':'Black', 'year':2021}
print(dict_1)
print(dict_1['other_attributes']['year'])
```

```python
dict_1['previous_owners']=['Mike', 'Shelly', 'Nitin']
print(dict_1)
```

```python
p = dict_1.pop('previous_owners')
print(p)
print(dict_1)
```

```python
print("Items: ", dict_1.items())
print("Keys: ", dict_1.keys())
print("Values: ", dict_1.values())
```

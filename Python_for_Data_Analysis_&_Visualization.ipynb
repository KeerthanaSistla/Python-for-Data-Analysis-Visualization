#Functions

```python
def welcome_message():
    print("Welcome")
welcome_message()
```

```python
def welcome(name):
    print(f"Welcome {name}")
welcome("Keerthana")
```

```python
def welcome(name1, name2, name3):
    print(f"Welcome {name1}, {name2} and {name3}")
welcome('A', 'B', 'C')
```

```python
def welcome(firstName, lastName):
    return f"Welcome {firstName} {lastName}"
print(welcome("Keerthana", "Sistla"))
```

```python
def greater_than_10(x):
    for ele in x:
        if ele > 10:
            print("True")
        else:
            print("False")

greater_than_10([1, 32, 10])
```

```python
def sum(l):
    s = 0
    for ele in l:
        s += ele
    return s

def avg(l):
    total = sum(l)
    length = len(l)
    return total / length

print(avg([1, 2, 3, 4, 5]))
```

```python
def my_func(*args):
    print(type(args))
    return args

print(my_func(1,2,3))
```

```python
def key_word_func(**kwargs):
    print(type(kwargs))
    return kwargs

print(key_word_func(name="K", age=21))
```

```python
def args_kwargs(*args, **kwargs):
    print(args)
    print(kwargs)

args_kwargs(1,2,3,4,"random_string",sport="football",age=33)
```

Map function

```python
countries_list = ['Argentina', "United Stats", "Germany"]

for x in countries_list:
    print(len(x))

print(list(map(len, countries_list)))
```

```python
def return_len(x):
    return f"{x}: {len(x)}"

print(return_len('Hello'))

print(list(map(return_len, countries_list)))
```

Filter Method

```python
countries_list = ['Argentina', 10, "Mexico", 18, "Brazil"]

def is_str(x):
    return type(x)==str

print(list(filter(is_str, countries_list)))
print(list(map(is_str, countries_list)))
```

Lambda Function

```python
print(lambda x: type(x)==str)
print(list(filter(lambda x: type(x)!=str, countries_list)))
```

```python
lst = [1,2,3,4,5]
sq = list(map(lambda x: x**2, lst))
print(sq)
```

#Introduction to Python

```python
a = 16
b = 3

print("Sum (a+b) =", a+b)
print("Difference (a-b) =", a-b)
print("Absolute Difference (abs(a-b)) =", abs(a-b))
print("Product (a*b) =", a*b)
print("Floor Division (a//b) =", a//b)
print("Quotient (round(a/b, 4)) =", round(a/b, 4))
print("Reminder (a%b) =", a%b)
print("Exponent (a**b) =", a**b)

x = input("Enter two numbers: ")
y = input()
print(f'x is {x}, y is {y}')
print("x==y:", x==y)
print("x!=y:", x!=y)
```

#String

```python
str = "This Is A String"
s2 = "This is a string \nwritten in two lines"

print(str)
print(s2)

print(str[:4:])
print(str[6:])
print(str[::3])
print(str[::-1])
print(str[13:2:-2])
```

Methods and Properties

```python
print(str.upper())
print(str.lower())

a = "Red Flag"
print("'a' is", a)
a.replace("Red", "Green")
print("'a' changed to", a)

b = "THIS, IS A, STRING !!!"
print("'b' is", b)
print(b.split())
print(b.split(","))
print(b.split("!"))
```

Concatination

```python
print("Concatination of strings 'a' and 'b'")
print(a + "... " + b)
```

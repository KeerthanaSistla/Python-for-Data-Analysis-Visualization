#Bar Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('top_10_countries.csv')
```

```python
df.rename(columns={'Country / Dependency':'Country'}, inplace=True)
```

```python
df
```

```python
df_g = df[['Region','Population']]
```

```python
df_g
```

```python
df_g = df_g.groupby(by='Region').sum()
```

```python
df_g.reset_index(inplace=True)
```

```python
df_g
```

```python
fig=plt.figure(figsize=(5,5))

# axes
axes = fig.add_axes([0,0,1,1])
axes.set_xlabel('Region')
axes.set_ylabel('Total Population')
axes.set_title('Top 10 Countries Population by Region')
```

```python
fig=plt.figure(figsize=(5,5))

# axes
axes = fig.add_axes([0,0,1,1])
axes.set_xlabel('Region')
axes.set_ylabel('Total Population')
axes.set_title('Top 10 Countries Population by Region')

# bar plot
plt.bar(x=df_g['Region'], height=df_g['Population'])

plt.show()
```

```python
fig=plt.figure(figsize=(5,5))

axes = fig.add_axes([0,0,1,1])
axes.set_xlabel('Region')
axes.set_ylabel('Total Population')
axes.set_title('Top 10 Countries Population by Region')

# bar plot
plt.bar(x=df_g['Region'], height=df_g['Population'], color=['b','r','y','g'], edgecolor='black')

plt.show()
```

#Line Plots

```python
import matplotlib.pyplot as plt
import pandas as pd
import datetime as dt
```

```python
tfl = pd.read_csv('tfl-daily-cycle-hires.csv')
```

```python
tfl.drop(columns='Unnamed: 2', inplace=True)
```

```python
tfl['Day'] = pd.to_datetime(tfl['Day'],format='%d/%m/%Y')
```

```python
tfl.dtypes
```

```python
tfl['Day'] = tfl['Day'].astype('datetime64[ns]')
```

```python
tfl.sort_values('Day', inplace=True)
```

```python
tfl['Year'] = tfl['Day'].dt.strftime('%Y')
```

```python
tfl_y = tfl[['Year', 'Number of Bicycle Hires']]
```

```python
tfl_y
```

```python
tfl_y = tfl_y.groupby(by='Year').sum().reset_index()
```

```python
tfl_y
```

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])
```

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])

ax1.plot(tfl_y['Year'],tfl_y['Number of Bicycle Hires'])

plt.show()
```

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])

ax1.plot('Year', 'Number of Bicycle Hires',data=tfl_y.tail(10), marker='1', markersize=20, linestyle=(0, (1, 10)))

plt.show()
```

```python
tfl_y.tail(10)
```

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])

ax1.plot(tfl_y['Year'].tail(10),tfl_y['Number of Bicycle Hires'].tail(10))

plt.show()
```

#Histogram

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
players = pd.read_csv('players_21.csv')
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])
ax1.set_ylabel('Number of Players')
ax1.set_xlabel('Overall score (bins)')

ax1.hist(x=players['overall'])

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])
ax1.set_ylabel('Number of Players')
ax1.set_xlabel('Overall score (bins)')

ax1.hist(x=players['overall'], bins=20)

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])
ax1.set_ylabel('Number of Players')
ax1.set_xlabel('Overall score (bins)')

ax1.hist(x=players['overall'], bins=[0,25,50,75,100],edgecolor='black')

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])
ax1.set_ylabel('Number of Players')
ax1.set_xlabel('Overall score (bins)')

plt.xticks([0,25,50,75,100],[0,25,50,75,100])

ax1.hist(x=players['overall'], bins=[0,25,50,75,100],edgecolor='black')

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])
ax1.set_xlabel('Number of Players')
ax1.set_ylabel('Overall score (bins)')

plt.yticks([0,25,50,75,100],[0,25,50,75,100])

ax1.hist(players['overall'], bins=[0,25,50,75,100], edgecolor='black',orientation='horizontal')

plt.show()
```

#Box Plots

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
players = pd.read_csv('players_21.csv')
```

```python
fig = plt.figure(figsize = (5,5))
ax1 = fig.add_axes([0,0,1,1])

ax1.boxplot(players['overall'], labels=['overall score'])

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])

ax1.boxplot(players['overall'], labels=['overall score'], showfliers=False)

plt.show()
```

```python
fig = plt.figure(figsize = (5,5))

ax1 = fig.add_axes([0,0,1,1])

ax1.boxplot(players['overall'], labels=['overall score'], showfliers=False, vert=False)

plt.show()
```

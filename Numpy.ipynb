#Numpy

```python
import numpy as np
```

```python
list1 = [1,2,3,4,5]
x = np.array(list1)
print(x)
print(type(x))
print(np.shape(x))
print(x.shape)
```

```python
print(type(np.array([2,2,2,2])))
print(type(np.array((3,3,3,3))))
```

2D Arrays

```python
arr2d = np.array([[1,1,1],[2,2,2],[3,3,3]])
print(arr2d)
```

Datatype Objects

```python
print(np.longdouble(100)/np.longdouble(3))
print(np.array([1,2,3.1453,5],dtype='g'))
```

Slicing and Indexing

```python
array_1d = np.array([2,5,6,7,6,4,3])
array_2d = np.array([[2,5,6,7,60,47,3],[11,50,6,9,62,42,30],[16,53,86,69,2,2,50]])
print(array_1d)
```

```python
print(array_1d[4])
print(array_1d[4:])
print(array_1d[1:6])
print(array_1d[::-1])
```

```python
print(array_2d)
print(array_2d[1:,2:])
print(array_2d[2,6])
print(array_2d[::-1,])
print(array_2d[::-1,])
```

Boolean Masking

```python
print(array_1d)
print(array_1d>3)
print(array_1d[array_1d>3])
```

```python
print(array_2d)
print(array_2d>10)
print(array_2d[array_2d>10])
```

Shape of Array

```python
print(array_2d[array_2d>10].shape)
print(array_2d[array_2d>10].reshape(2,6))
```

```python
array_1 = np.array([1,2,3,4,5,6,4,2,2,1])
print(array_1)
np.reshape(array_1,(2,5))
print(array_1)
array_1.flatten().shape
print(array_1)
```

```python
array_2 = np.array([[2,3,2,4,6],[5,6,3,2,4]])
print(array_2)
print(np.ravel(array_2))
```

Concatinate

```python
array_1 = np.array([[1, 1, 1], [2, 2, 2]])
array_2 = np.array([[3, 3, 3],[4, 4, 4],[5, 5, 5]])
array_3 = np.array([[3, 3],[4, 4],[5, 5]])
array_4 = np.concatenate((array_1,array_2),axis=0)
print(array_4)
array_5 = np.concatenate((array_3,array_2),axis=1)
print(array_5)
```

Transpose Array

```python
print(array_1)
print(np.transpose(array_1))
```

Sorting Elements

```python
array_unsort = np.array([[1,10,3],[2,3,1]])
print(array_unsort)
print(np.sort(array_unsort))
print(np.sort(array_unsort,axis=1))
print(np.sort(array_unsort,axis=0))
```

Array Arithmatic (Broadcasting)

```python
array_1 = np.array([1, 2, 3])
array_2 = np.array([5, 6, 7])
array_3 = np.array([[1, 2, 3, 4, 5], [5, 3, 2, 4, 2], [1, 2, 3, 4, 5]])
```

```python
print(array_1+array_2)
print(np.add(array_1,array_2))
print(array_1-array_2)
print(np.subtract(array_1,array_2))
```

```python
print(array_1+5)
print(array_1-5)
print(array_1*5)
print(array_1/5)
```

Sum of elements

```python
print(array_3)
print(sum(array_3))
```

Cumilative Sum of Elements in an Axis

```python
print(np.cumsum(array_3))
print(np.cumsum(array_3, axis=0))
print(np.cumsum(array_3, axis=1))
```

Arg Max/Min - Max/Min in given axis

```python
print(np.amax(array_3, axis=0))
print(np.amin(array_3, axis=0))
```

Average of elements in given axis

```python
print(np.mean(array_3, axis=0))
```

Using External Database

```python
x = np.loadtxt("tfl-daily-cycle-hires.txt",delimiter=',',usecols=(1),skiprows=1)
print("Number of Zeros:", np.count_nonzero(x))
print("Max:", x.max())
print("Min:", x.min())
print("Mean:", x.mean())
print("Std:", x.std())
```

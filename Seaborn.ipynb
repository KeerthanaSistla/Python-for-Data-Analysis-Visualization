```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#Bar Plots

```python
sns.barplot(x = countries_df['Region'], y=countries_df['Population'])
```

```python
sns.barplot(x = countries_df['Region'], y=countries_df['Population'], ci=None)
```

```python
sns.barplot(y = countries_df['Region'], x=countries_df['Population'], ci=None, orient="h")
```

```python
sns.barplot(x = 'Region', y='Population', data=countries_df, ci=None, estimator=sum)
plt.show()
```

```python
sns.barplot(x = 'Region', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region', dodge=False)
plt.show()
```

```python
plt.figure(figsize=(10,10))

sns.barplot(x = 'Country', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region')

plt.legend(loc='center')

plt.show()
```

```python
plt.figure(figsize=(10,10))

sns.barplot(x = 'Country', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region', dodge=False)

plt.legend(loc='center')

plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(figsize=(20,10),nrows=1,ncols=2, tight_layout=True)

sns.barplot(x=countries_df['Region'],y=countries_df['Population'], ci=None, estimator=sum, ax=ax1)
sns.barplot(x=countries_df['Country'],y=countries_df['Population'], ci=None, estimator=sum, ax=ax2)

plt.show()
```

#Matrix Plots

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
players = pd.read_csv('players_21.csv')
```

```python
pd.set_option('display.max_columns', False)
```

```python
players = players[~players['team_position'].isin(['RES', 'SUB'])]
```

```python
players.head(1)
```

#Heatmaps

```python
player_matrix = pd.pivot_table(data=players, index='overall', columns='team_position', values='sofifa_id',aggfunc='count',fill_value=0 )
```

```python
player_matrix.sort_values('overall', ascending=False, inplace=True)
```

```python
player_matrix.head()
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(player_matrix)
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(player_matrix,cmap='YlGnBu')
```

#Clustermaps

```python
plt.figure(figsize=(10,10))
sns.clustermap(player_matrix,cmap='YlGnBu')
```

#Multi Plot Grids

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.head()
```

#FacetGrid

```python
sns.relplot(data=penguins, x= 'body_mass_g', y='flipper_length_mm', row='island', col='species', kind='scatter', height=3)
```

```python
g = sns.FacetGrid(data=penguins, col='island',row='species')
g.map_dataframe(sns.scatterplot,x='body_mass_g', y='flipper_length_mm')
```

#Pair Plots

```python
sns.pairplot(penguins)
```

```python
sns.pairplot(penguins, kind='kde')
```

```python
sns.pairplot(penguins, diag_kind='kde')
```

```python
sns.pairplot(penguins, kind='kde', corner=True, hue='species')
```

#Pair Grids

```python
g=sns.PairGrid(penguins)
g.map(sns.scatterplot)
```

```python
g=sns.PairGrid(penguins)
g.map(sns.histplot)
```

```python
g = sns.PairGrid(penguins)
g.map_lower(sns.kdeplot)
g.map_diag(sns.histplot)
g.map_upper(sns.histplot)
```

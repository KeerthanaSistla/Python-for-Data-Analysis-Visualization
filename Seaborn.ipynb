#Categorical Plots

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
countries_df = pd.read_csv('top_10_countries.csv')
```

```python
countries_df
```

#Bar Plots

```python
sns.barplot(x = countries_df['Region'], y=countries_df['Population'])
```

```python
sns.barplot(x = countries_df['Region'], y=countries_df['Population'], ci=None)
```

```python
sns.barplot(y = countries_df['Region'], x=countries_df['Population'], ci=None, orient="h")
```

```python
sns.barplot(x = 'Region', y='Population', data=countries_df, ci=None)
```

```python
sns.barplot(x = 'Region', y='Population', data=countries_df, ci=None, estimator=sum)
plt.show()
```

```python
sns.barplot(x = 'Region', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region', dodge=False)
plt.show()
```

```python
plt.figure(figsize=(10,10))

sns.barplot(x = 'Country', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region')

plt.legend(loc='center')

plt.show()
```

```python
plt.figure(figsize=(10,10))

sns.barplot(x = 'Country', y='Population', data=countries_df, estimator=sum, ci=None, hue='Region', dodge=False)

plt.legend(loc='center')

plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(figsize=(20,10),nrows=1,ncols=2, tight_layout=True)

sns.barplot(x=countries_df['Region'],y=countries_df['Population'], ci=None, estimator=sum, ax=ax1)
sns.barplot(x=countries_df['Country'],y=countries_df['Population'], ci=None, estimator=sum, ax=ax2)

plt.show()
```

Relational Plots

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import datetime as dt
```

```python
stocks = pd.read_excel('stocks.xlsx')
```

```python
stocks.head()
```

```python
stocks['Year']=stocks['Date'].dt.strftime('%Y')
```

```python
sns.set_theme()
```

#Line Plots

```python
sns.lineplot(x='Date', y='Close', data=stocks, hue='Symbol')
```

```python
plt.figure(figsize=(20,10))

sns.lineplot(x='Date', y='Close', data=stocks, hue='Symbol', linestyle='--')
```

```python
plt.figure(figsize=(20,10))

sns.lineplot(x='Date', y='Close', data=stocks, hue='Symbol', linestyle='--', palette='Set2')
```

```python
plt.figure(figsize=(20,10))

sns.lineplot(x='Date', y='Close', data=stocks, hue='Symbol', linestyle='--', palette=['#707070','#7cbb00', '#3b5998'])
```

```python
fig, (ax1,ax2,ax3) = plt.subplots(1,3, figsize=(20,5))

sns.lineplot(x='Date', y='Close', data=stocks[stocks['Symbol']=='AAPL'],ax=ax1)
sns.lineplot(x='Date', y='Close', data=stocks[stocks['Symbol']=='FB'],ax=ax2)
sns.lineplot(x='Date', y='Close', data=stocks[stocks['Symbol']=='MSFT'],ax=ax3)
```

#Distribution Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.head()
```

#Rug Plots

```python
sns.rugplot(x='body_mass_g', data=penguins)
```

```python
sns.rugplot(x='body_mass_g', height=2, data=penguins)
```

```python
fig=plt.figure(figsize=(5,5))

ax = fig.add_axes([0,0,1,1])

ax.set_ylim(165,235)

sns.rugplot(x='body_mass_g',y='flipper_length_mm',
            data=penguins)
```

```python
fig=plt.figure(figsize=(5,5))
ax=fig.add_axes([0,0,1,1])

ax.set_ylim(165,235)

sns.rugplot(x='body_mass_g',y='flipper_length_mm', data=penguins)
sns.scatterplot(x='body_mass_g',y='flipper_length_mm', data=penguins)
```

#Hist Plots

```python
sns.histplot(x='body_mass_g', data=penguins)
```

```python
sns.histplot(x='body_mass_g', data=penguins, hue='island')
```

```python
sns.histplot(x='body_mass_g', data=penguins, y='island')
```

#KDE Plots

```python
sns.kdeplot(x='body_mass_g', data = penguins)
```

```python
sns.kdeplot(x='body_mass_g', data=penguins)
sns.rugplot(x='body_mass_g', data=penguins)
```

```python
sns.kdeplot(x='body_mass_g', y = 'flipper_length_mm', data = penguins)
```

```python
sns.kdeplot(x='body_mass_g',y='flipper_length_mm' ,data=penguins, fill=True)
sns.scatterplot(x='body_mass_g',y='flipper_length_mm' ,data=penguins)
```

#EDCF Plots

```python
sns.ecdfplot(x='flipper_length_mm', data = penguins)
```

```python
sns.ecdfplot(y='flipper_length_mm', data = penguins)
```

```python
sns.ecdfplot(y='flipper_length_mm', data=penguins, hue='island')

plt.grid()
```
